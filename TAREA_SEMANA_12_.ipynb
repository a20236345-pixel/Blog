```python
# Importo pandas para leer el Excel y folium para crear el mapa.
import pandas as pd
import folium

# Leo el archivo .xlsx (antes lo subo a Colab con el ícono de la carpeta).
datos = pd.read_excel("f1_2016.xlsx")

# Muestro las primeras filas para comprobar que se cargó bien.
print(datos.head())

# Ahora hago una lista con los dos pilotos que elegí.
pilotos_elegidos = ["S. VETTEL", "F. ALONSO"]

# Con isin() me quedaré solo con las filas cuyo piloto está en mi lista.
seleccion = datos[datos["Piloto"].isin(pilotos_elegidos)].copy()
print(seleccion)

# Ahora sumo los puntos de todas las carreras de cada piloto.
# iloc[:, 1:] toma todas las columnas menos la primera (que es el nombre)
# y sum(axis=1) suma los valores de cada fila.
seleccion["Total"] = seleccion.iloc[:, 1:].sum(axis=1)
print(seleccion[["Piloto", "Total"]])

# Creo un diccionario donde cada piloto tiene su escudería en 2016,
# el enlace de su foto y las coordenadas de su ciudad de nacimiento.
# Vettel nació en Heppenheim (Alemania) y Alonso en Oviedo (España).
info_pilotos = {
    "S. VETTEL": {
        "escuderia": "Scuderia Ferrari",
        "foto": "PEGA_AQUI_EL_ENLACE_DE_LA_FOTO_DE_VETTEL",
        "latitud": 49.64,
        "longitud": 8.64
    },
    "F. ALONSO": {
        "escuderia": "McLaren-Honda",
        "foto": "PEGA_AQUI_EL_ENLACE_DE_LA_FOTO_DE_ALONSO",
        "latitud": 43.3614,
        "longitud": -5.8593
    }
}

# Ahora guardo el color del marcador de cada piloto (rojo Ferrari y naranja McLaren).
colores = {"S. VETTEL": "red", "F. ALONSO": "orange"}

# 5.1 Mapa base: pongo el centro entre las dos ciudades y un zoom de 4 para que se vean ambos puntos al mismo tiempo. En tiles uso el mapa de OpenStreetMap y en attr pongo los créditos (licencia) que exige OpenStreetMap para poder usar su mapa.
mapa = folium.Map(
    location=[46.5, 1.5],
    zoom_start=4,
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}",
    attr="Tiles &copy; Esri &mdash; Source: Esri, DeLorme, NAVTEQ, USGS, and others"
)

# 5.2 Recorro cada piloto de mi selección para crear su marcador.
for i, fila in seleccion.iterrows():
    nombre = fila["Piloto"]
    puntaje = fila["Total"]
    datos_piloto = info_pilotos[nombre]

    # 5.3 Creo el contenido de la ventana emergente (popup) con etiquetas HTML:
    # <h2> para el nombre, <img> para la foto, <p> para cada dato,
    # <b> para resaltar los títulos y <i> para el puntaje, y <div> para agrupar todo.
    html = f"""
    <div style="text-align:center; width:220px;">
        <h2>{nombre}</h2>
        <img src="{datos_piloto['foto']}" width="150">
        <p><b>Escudería:</b> {datos_piloto['escuderia']}</p>
        <p><b>Puntaje total 2016:</b> <i>{puntaje} puntos</i></p>
    </div>
    """
    popup = folium.Popup(html, max_width=250)

    # 5.4 Creo el marcador en la ciudad de nacimiento, con ícono de auto
    # y color personalizado. tooltip muestra el nombre al pasar el mouse.
    marcador = folium.Marker(
        location=[datos_piloto["latitud"], datos_piloto["longitud"]],
        popup=popup,
        tooltip=nombre,
        icon=folium.Icon(color=colores[nombre], icon="car", prefix="fa")
    )

    # 5.5 Agrego el marcador al mapa.
    marcador.add_to(mapa)

# Finalmente muestro el mapa con display().
from IPython.display import display
display(mapa)

# Ahora imprimo la justificación debajo del mapa.
print("Organicé la ventana emergente en un solo bloque <div> centrado para que toda la información del piloto aparezca agrupada y ordenada, con un ancho fijo que evita que se vea desproporcionada. Usé <h2> para el nombre porque es lo más importante y debe destacar como subtítulo sin ser tan grande como un título principal. Debajo puse la fotografía con <img> en un tamaño pequeño (150 px), para identificar visualmente al piloto sin que la ventana ocupe demasiado espacio en el mapa. Para los datos usé párrafos <p>, de modo que cada dato quede en su propia línea, con la etiqueta en negrita <b> para que se distinga de su valor, y el puntaje en cursiva <i> para resaltar la cifra. El orden va de lo general a lo específico: primero quién es el piloto, luego su imagen y al final sus datos de la temporada 2016 (escudería y puntaje), lo que facilitará una lectura rápida.")
```
