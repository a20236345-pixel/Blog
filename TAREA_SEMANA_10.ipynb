```python
# Importamos el archivo cargado anteriormente
with open("Borderline sagrada cap 1.txt", "r") as file:
    capitulo = file.read()
```

```python
# Muestro el contenido original para comprobar que se cargó correctamente.
print("TEXTO ORIGINAL")
print(capitulo)

# Convierto todo a minúsculas con lower()
texto_min = capitulo.lower()

# Con replace() cambio cada signo por "". Identifiqué que estos son los signos que aparecen en este capítulo: . , ; : « »
texto_depurado = texto_min.replace(".", "").replace(",", "").replace(";", "").replace(":", "").replace("«", "").replace("»", "")

# Con split() separo el texto en palabras y obtengo una lista.
lista_palabras = texto_depurado.split()
print("LISTA DE PALABRAS")
print(lista_palabras)
print("Total de palabras:", len(lista_palabras))

# Estas son las palabras que no deseo analizar: artículos, preposiciones, conjunciones, pronombres, adverbios muy comunes y verbos auxiliares.
stopwords = ["a", "al", "ante", "con", "contra", "de", "del", "desde", "en", "entre",
             "hacia", "hasta", "para", "por", "sin", "sobre", "tras",
             "el", "la", "los", "las", "un", "una", "unos", "unas", "lo",
             "y", "e", "o", "u", "ni", "pero", "que", "como", "si", "sino",
             "aunque", "porque", "cuando", "mientras",
             "yo", "él", "ella", "ellos", "ellas", "me", "te", "se", "nos",
             "le", "les", "mí", "qué", "quién",
             "mi", "mis", "tu", "su", "sus", "ese", "esa", "eso", "este", "esta", "esto",
             "no", "muy", "más", "tan", "tanto", "tanta", "ya", "así", "también",
             "solo", "nada", "todo", "todos", "todas", "mismo", "siempre", "mucho", "mucha",
             "es", "son", "era", "fue", "ser", "estar", "estoy", "estaba", "está",
             "voy", "va", "tengo", "tenía", "hay", "habría", "hubiese", "había",
             "hacía", "hacer", "ha", "he", "han", "hemos", "seguido",
             "veces", "vez", "cada", "otro", "otra", "ahora", "hoy"]

# Creo una lista vacía y recorro lista_palabras con un for. Guardo la palabra solo si no está en stopwords. Conservo los números (2011, 13) porque las indicaciones solo piden eliminar signos de puntuación y stopwords, y porque las fechas también forman parte del contenido del texto.
palabras_limpias = []

for palabra in lista_palabras:
    if palabra not in stopwords:
        palabras_limpias.append(palabra)

print("LISTA SIN STOPWORDS")
print(palabras_limpias)
print("Palabras que quedaron:", len(palabras_limpias))

# Creo un diccionario vacío para guardar cada palabra y cuántas veces aparece en la lista depurada.
frecuencias = dict()

for palabra in palabras_limpias:
    if palabra in frecuencias:
        # Si ya existe, le sumo 1
        frecuencias[palabra] = frecuencias[palabra] + 1
    else:
        # Si es la primera vez que aparece, empieza en 1
        frecuencias[palabra] = 1

print("DICCIONARIO DE FRECUENCIAS")
print(frecuencias)
print("Palabras únicas:", len(frecuencias))

# Creo la nube con WordCloud y la personalizo:
#   background_color -> color de fondo
#   colormap         -> paleta de colores de las palabras
#   max_words        -> máximo de palabras que se dibujan
# generate_from_frequencies() usa directamente mi diccionario de frecuencias.
nube = WordCloud(width=900,
                 height=500,
                 background_color="black",
                 colormap="plasma",
                 max_words=100).generate_from_frequencies(frecuencias)

# Ahora dibujo la nube con matplotlib
plt.figure(figsize=(10, 6))
plt.imshow(nube, interpolation="bilinear")
plt.title("Nube de palabras - Capítulo 1")

# Oculto los ejes porque en una nube de palabras no aportan información
plt.axis("off")
plt.show()
```
