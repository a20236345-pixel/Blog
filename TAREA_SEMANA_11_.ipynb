```python
# Importo pandas para leer el .csv y matplotlib para hacer los gráficos.
import pandas as pd
import matplotlib.pyplot as plt

# Leo el archivo .csv (primero hay que subirlo a Colab con el ícono de la carpeta)
datos = pd.read_csv("SP1_2025_2026.csv")

# Muestro las primeras filas para comprobar que se cargó bien
print(datos[["HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR", "HY", "AY", "HR", "AR"]].head())

# Coloco la variable para el equipo elegido
equipo = "Oviedo"

# Filtro los partidos donde el Oviedo jugó como LOCAL (columna HomeTeam)
local = datos[datos["HomeTeam"] == equipo]

# Filtro los partidos donde el Oviedo jugó como VISITANTE (columna AwayTeam)
visitante = datos[datos["AwayTeam"] == equipo]

print("Partidos del Oviedo como local:", len(local))
print("Partidos del Oviedo como visitante:", len(visitante))

# Colco los goles anotados por el Oviedo como local = FTHG (goles del equipo local)
goles_anotados_local = local["FTHG"]

# Goles recibidos por el Oviedo como visitante = FTHG (goles del rival, que es el local)
goles_recibidos_visitante = visitante["FTHG"]

# Creo una figura grande y pongo los dos histogramas lado a lado
plt.figure(figsize=(12, 5))

# Histograma 1: goles anotados como local
plt.subplot(1, 2, 1)
# bins=range(0, 6) y align="left" hacen que cada barra quede centrada en un número de goles
plt.hist(goles_anotados_local, bins=range(0, 6), align="left", rwidth=0.8, color="royalblue", edgecolor="black")
plt.title("Goles anotados por el Oviedo como local")
plt.xlabel("Goles por partido")
plt.ylabel("Cantidad de partidos")
plt.xticks(range(0, 5))

# Histograma 2: goles recibidos como visitante
plt.subplot(1, 2, 2)
plt.hist(goles_recibidos_visitante, bins=range(0, 6), align="left", rwidth=0.8, color="tomato", edgecolor="black")
plt.title("Goles recibidos por el Oviedo como visitante")
plt.xlabel("Goles por partido")
plt.ylabel("Cantidad de partidos")
plt.xticks(range(0, 5))

# Ajusto los espacios para que no se junten los títulos y muestro los gráficos
plt.tight_layout()
plt.show()

# Interpretación de los histogramas
print("Como local, el Oviedo anotó muy poco, pues no marcó en 10 de sus 19 partidos y solo hizo 1 gol en los otros 9, sin superar nunca un gol por partido (9 goles en total). Como visitante, en cambio, recibió goles con mucha frecuencia, debido a que lo más común fue recibir 3 goles (7 partidos), seguido de 1 gol (5 partidos) y 2 goles (4 partidos). Solo en un partido de visita no recibió goles, y en dos recibió 4. En total recibió 42 goles de visitante, con un promedio de unos 2.2 por partido, mientras que de local anotó un promedio de unos 0.5. La comparación muestra dos tendencias: un ataque poco productivo en casa y una defensa muy frágil fuera de casa. La distribución de goles anotados se concentra en valores bajos, mientras que la de goles recibidos está más dispersa y desplazada hacia valores altos.")


# En los partidos como local, la columna FTR significa:H = ganó el Oviedo, D = empate, A = perdió el Oviedo
ganados = len(local[local["FTR"] == "H"])
empatados = len(local[local["FTR"] == "D"])
perdidos = len(local[local["FTR"] == "A"])

print("Ganados:", ganados, "| Empatados:", empatados, "| Perdidos:", perdidos)

plt.figure(figsize=(6, 6))
plt.pie([ganados, empatados, perdidos],
        labels=["Ganados", "Empatados", "Perdidos"],   # etiquetas de cada categoría
        colors=["mediumseagreen", "gold", "indianred"], # colores de las porciones
        explode=[0.05, 0.05, 0.05],                     # separo un poco cada porción
        shadow=True,                                    # agrego sombra
        autopct="%1.1f%%",                              # muestro el porcentaje
        startangle=90)
plt.title("Resultados del Oviedo jugando como local")
plt.show()

# Interpretación del gráfico de pastel
print("De los 19 partidos que el Oviedo jugó como local, perdió 8 (42.1%), empató 7 (36.8%) y ganó solo 4 (21.1%). Es decir, casi 8 de cada 10 partidos en casa terminaron sin victoria. Los empates fueron la segunda categoría más frecuente, lo que muestra que el equipo logró sacar al menos un punto en varios partidos. Aun así, jugar de local no le dio una ventaja clara, porque las derrotas fueron el resultado más común.")

# Agrupo por AwayTeam (equipo visitante), calculo el promedio de AY (amarillas) y redondeo al entero más cercano con round()
promedio_amarillas = datos.groupby("AwayTeam")["AY"].mean().round().astype(int)
print(promedio_amarillas)

plt.figure(figsize=(8, 7))
# barh() hace las barras horizontales
plt.barh(promedio_amarillas.index, promedio_amarillas.values, color="darkorange", edgecolor="black")
plt.title("Promedio de tarjetas amarillas por equipo jugando como visitante")
plt.xlabel("Promedio de tarjetas amarillas (redondeado)")
plt.ylabel("Equipo")
plt.yticks(rotation=0)   # dejo los nombres de los equipos horizontales para leerlos mejor
plt.tight_layout()
plt.show()

# Interpretación del gráfico de barras
print("Al redondear, todos los equipos quedan entre 2 y 3 tarjetas amarillas promedio como visitantes, por lo que las diferencias se ven pequeñas. Siete equipos llegan a 3 (Alavés, Espanyol, Getafe, Sevilla, Real Sociedad, Rayo Vallecano y Villarreal), y los otros trece, incluido el Oviedo, tienen 2. El Oviedo se ubica en el grupo de menor promedio, así que como visitante no es de los equipos más amonestados. El redondeo oculta diferencias como por ejemplo entre Getafe, que tiene el promedio más alto, y el Real Madrid, que tiene el más bajo, y por eso ambos aparecen con valores cercanos en el gráfico.")
```
